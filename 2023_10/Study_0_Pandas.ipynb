```python
exam_data = {'이름': ['서준','우현','인아'],
             '수학': [90,80,70],
             '영어': [95,75,60],
             '음악': [77,88,99],
             '체육': [90,100,90],}
```

```python
# '이름' column 을 새로운 index 로 지정
df_exam = pd.DataFrame(exam_data)
df_exam.set_index('이름',inplace=True)
```

```python
df_exam.loc['서준','음악']
```

```python
# Same Result & Different Way
# 1
c = df_exam.loc['서준',['음악','체육']]
print(c)
# 2
d = df_exam.iloc[0,[2,3]]
print(d)
# 3
e = df_exam.loc['서준','음악':'체육']
print(e)
# 4
f = df_exam.iloc[0,2:]
print(f)
```

```python
# Add Column
df_exam['국어'] = 80
print(df_exam)
```

```python
# Add Row
df_exam.loc['동규'] = [65,85,75,77,98]
print(df_exam)
```

```python
# Transpose
df_exam_T = df_exam.transpose()
print(df_exam_T)
```

```python
df_exam_T.index
```

```python
# Change index
df_exam_index = df_exam_T.set_index(['서준'])
print(df_exam_index)
```

```python
# Reset index
df_exam_R_index = df_exam_index.reset_index()
print(df_exam_R_index)
```

```python
# Sort Values
df_exam_sort = df_exam.sort_values(by='영어', ascending=False)
print(df_exam_sort)
```

# (3) 산술 연산

```python
# Divide
student1 = pd.Series({'국어':100,'영어':80,'수학':90})
print(student1)
print('\n')
student1_d_200 = student1 / 200
print(student1_d_200)
```

```python
# Series + add + Series
student2 = pd.Series({'수학':90,'국어':85,'영어':75})

addition = student1 + student2
subtraction = student1 - student2

# Series to DataFrame
result = pd.DataFrame([addition,subtraction],index=['덧셈','뺄셈'])
print(result)
```

```python
# Using Method
student3 = pd.Series({'국어':np.nan,'영어':88,'수학':99})

sr_add = student1.add(student3, fill_value=0)
print(sr_add)
```
